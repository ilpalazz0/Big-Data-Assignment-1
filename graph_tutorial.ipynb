```r
if (!require("igraph")) install.packages("igraph")
library(Matrix)
library(igraph)
library(sna)
```

```r
file_path <- "/Users/elmarmammadov/Documents/BigData/p1/fb-pages-artist/fb-pages-artist.edges"
edges <- read.table(file_path, header = FALSE, sep = ",", stringsAsFactors = FALSE)
```

```r
g <- graph_from_data_frame(edges, directed = FALSE)
```

```r
V(g)
E(g)
```

```r
g.adj <- igraph::get.adjacency(g)
g.adj
```

```r
g.density <- igraph::edge_density(g)
g.density
```
